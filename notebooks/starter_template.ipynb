```python
%%sql
-- Aufgabe 1
Select count(l_returnflag),
        r_name
from lineitem
join orders on o_orderkey=l_orderkey
join customer on c_custkey = o_custkey
join nation on n_nationkey =c_nationkey
join region on r_regionkey = n_regionkey
where l_returnflag = "R"
group by r_name
```

```python
%%sql
-- Aufgabe 2 List the top 10 most selling parts (part name)
select 
    p_name,
    sum(l_quantity) as sold_quantity
from lineitem
left join part on p_partkey = l_partkey
group by p_name
order by sold_quantity desc
limit 10
```

```python
%%sql
-- Aufgabe 2 List the top 10 most selling parts (part name)
select 
    p_name
from lineitem
left join part on p_partkey = l_partkey
where l_returnflag = "N"
group by p_name
order by sum(l_quantity) desc
limit 10
```

```python
%%sql
-- Aufgabe 3 Sellers (name) who have sold at least one of the top 10 selling parts
select 
    s_name,
    count(p_name)
from partsupp
left join part on p_partkey=ps_partkey
left join supplier on s_suppkey = ps_suppkey
where p_name in (select 
    p_name
from lineitem
left join part on p_partkey = l_partkey
group by p_name
order by sum(l_quantity) desc
limit 10)
group by s_name


```
